```python
# Write a python program to convert a string to title case without using the tit

s=input("Enter a String :")
x=""
l=s.split()

for i in l:
    x=x+i[0].upper()+i[1::]+" "
    
x=x.strip()
print(f"String to Title : {x}")
```

```python
"""
Write a program to check if two strings are balanced. For example,
strings s1 and s2 are balanced if all the characters in the s1 are present in s2
The character’s position doesn’t matter.
"""

s1=input("enter string: ")
s2= input("enter string: ")
flag=0
for i in s1:
    if i in s2:
        continue
    else :
        flag=1
if(flag==1):
    print("Not Balanced")
else :
    print("Balanced")
    

```
