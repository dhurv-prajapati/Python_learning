```python
#write a function called allevens that given a list of integers,returns even list 

def allevens(lst):
    my_list=[]

    for i in lst:
        if(i%2==0):
            my_list.append(i)
    return my_list
    
l=eval(input("Enter the list of intergers only with [ ]"))
print(f"Even list:{allevens(l)}")
```

```python
"""Create a function that will return another string similar to the input
example, input of “Mr. Ed” will result in “mR. eD” as the output string
function is prohibited."""

def case_invert(s):
    s1=""

    for i in s :
        if(i.isupper()):
            s1=s1+i.lower()
        elif(i.islower()):
            s1=s1+i.upper()
        else :
            s1=s1+i
    return s1

x=case_invert(input("Enter String:"))
print(x)

```
