```python
# Product of Odd's in particular number 
# n%2==0 --> Even
# n%2!=0 --> Odd
# to find last digit of the number Number%10=last digit 
# for product calculation we will initialised our var with 1 (for sumation 0)
num=int(input("Enter a number "))
product=1

while num>0 :
    d=0
    d=(num)%10 

    if(d%2!=0):
        product=product*d
    num=num//10

print(f"Product of odd's ={product}")
```

```python
#  program to check if year is a leap year or not 
y=int(input("Enter a Year "))
if(y%4==0):
    if(y%400==0 or y%100!=0):
        print(f"{y} is Leap year!")
    else :
        print(f"{y} is Not Leap year!")
```

```python
# program to find average of first N natural numbers given by user.
# here we need Natural numbers than the range is (1,n+1)
# Avrage = sum//count of the numbers 
n=int(input("Enter a number "))
s=0
c=0

for i in range(1,n+1):
    s=s+i
    c=c+1

print(f"Sum ={s}\nAvarage={s//c}")
```

```python
# program to read three numbers (a,b,c) and check how many numbers between ‘a’ and ‘b’ are divisible by ‘c’.
a=int(input("Enter a number "))
b=int(input("Enter a number "))
c=int(input("Enter a number "))
count=0
for i in range(a,b+1,1):
    if(i%c==0):
        count=count+1

print(f"Count={count}")
```

```python
# program to find the factorial of a number provided by the user

n=int(input("Enter a number "))
fac=1
for i in range(n,0,-1):
    fac=fac*i

print(f"{n}! =",fac)
    
```

```python
# program to reverse a number
# 1234 ==> 4321 
n=int(input("Enter a number "))
rev=0
while n>0 :
    
    d=(n)%10
    rev=(rev*10)+(d)
    n=n//10

print("reverse = ",rev)

```
