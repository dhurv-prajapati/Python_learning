```python
# A water storage facility needs to store n liters of water using bottles of different sizes. The available bottle sizes are: 500 
# liters, 200 liters, 50 liters, 10 liters, 1 liter. 
# Task: Using the largest bottles first, calculate:
# Number of 500L bottles required
# Number of 200L bottles required
# Number of 50L bottles required
# Number of 10L bottles required
# Number of 1L bottles required

w=int(input("Enter water liter"))
print((w//500),end="")
r1=w%500
print(r1//200,end="")

r2=(r1%200)
print(r2//50,end="")

r3=r2%50
print(r3//10,end="")

r4=r3%10
print(r4//1,end="")
```
